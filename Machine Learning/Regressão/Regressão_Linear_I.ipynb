### **Regressão Linear Simples**

```python
X = df[["curbweight"]]
y = df["citympg"]
```

```python
reg_model = sm.OLS(y, sm.add_constant(X)).fit()
```

```python
print(reg_model.summary())
```

```python
X = df[["peakrpm"]]
y = df["citympg"]
```

```python
reg_model = sm.OLS(y, sm.add_constant(X)).fit()
```

```python
print(reg_model.summary())
```
