```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/FIAP/Machine Learning & Data Science/Datasets/dataset_armageddon_final_corrected.csv")
```

```python
df.describe(include='object')
```

```python
# verifica dados ausentes
print(df.isnull().sum() /len(df))
```

```python
num_cols = df.select_dtypes(include=['float64', 'int64']).columns # Seleciona as colunas com tipo de dados inteiro e float
plt.figure(figsize=(12,6)) # cria quadro
df[num_cols].hist(bins=30, figsize=(15,10), color='skyblue', edgecolor='black') # cria o histograma
plt.suptitle('Distribuição das Variáveis Numéricas', fontsize=16) # cria o titulo
plt.tight_layout() # ajusta o layout
plt.show()
```

```python
cat_cols = df.select_dtypes(include=['object']).columns # Seleciona as colunas com tipo de dados objeto
for col in cat_cols:
    plt.figure(figsize=(10,4)) # cria quadro
    sns.countplot(y=df[col], order=df[col].value_counts().index, palette='viridis') # cria o histograma
    plt.title(f'Distribuição da Variável : {col}', fontsize=14) # cria o titulo
    plt.show()
```

```python
#outliers
plt.figure(figsize=(12,6))
for col in num_cols:
    plt.figure(figsize=(10,4))
    sns.boxplot(y=df[col], color='lightcoral')
    plt.title(f"Boxplot da Variável {col}")
    plt.show()
```

```python
# Correlação entre os dados

df_corr = df[num_cols]
```

```python
df_corr.corr()
```

```python
plt.figure(figsize=(12,6)) # gera o quadro branco
corr_matrix = df_corr.corr() # gera a matriz de correlação
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5) #Heatmap, o grafico que demonstra a correlação
plt.title('Matriz de Correlação')
plt.show()
```

```python
# analise bivariada - Relação entre variáveis numéricas

for i in range(len(num_cols)):
    for j in range(i + 1, len(num_cols)):
        plt.figure(figsize=(8,5))
        sns.scatterplot(x=df[num_cols[i]], y=df[num_cols[j]], alpha=0.6, color="blue")
        plt.title(f"Relação entre {num_cols[i]} e {num_cols[j]}")
        plt.xlabel(num_cols[i])
        plt.ylabel(num_cols[j])
        plt.show()
```

```python
# Análise bivariada - Boxplot entre variáveis categoricas e numéricas
for cat in cat_cols:
    for num in num_cols:
        plt.figure(figsize=(10,5))
        sns.boxplot(x=df[cat], y=df[num], palette="Set2")
        plt.title(f"Boxplot de {num} por {cat}")
        plt.xticks(rotation=45)
        plt.xlabel(cat)
        plt.ylabel(num)
        plt.show()
```

```python
# Recarregando as bibliotecas após o reset do ambiente
import pandas as pd

# Criar DataFrame base para a aba "database"
database_df = pd.DataFrame(columns=["Data", "Descrição", "Valor", "Categoria", "Parcelas"])

# Criar DataFrame base para a aba "Resumo Mensal"
meses = pd.date_range(start=pd.to_datetime("today").replace(day=1), periods=12, freq='MS').strftime("%m/%Y")
resumo_df = pd.DataFrame(columns=["Mês", "Receitas", "Custos Fixos", "Custos Variáveis", "Investimentos", "Saldo Final"])
resumo_df["Mês"] = meses

# Criar DataFrame base para a aba "Projeção"
projecao_df = pd.DataFrame(columns=["Despesa"] + list(meses))

# Caminho do arquivo
file_path = "controle_financeiro_template.xlsx"

# Criando o arquivo Excel com fórmulas
with pd.ExcelWriter(file_path, engine="xlsxwriter") as writer:
    database_df.to_excel(writer, sheet_name="database", index=False)
    resumo_df.to_excel(writer, sheet_name="Resumo Mensal", index=False)
    projecao_df.to_excel(writer, sheet_name="Projeção", index=False)

    # Adicionar fórmulas no resumo mensal
    workbook = writer.book
    worksheet = writer.sheets["Resumo Mensal"]

    # Fórmulas para calcular saldo final (Receitas - Custos Fixos - Custos Variáveis - Investimentos)
    for i in range(2, 14):  # Para os 12 meses (linha começa na segunda linha do Excel)
        worksheet.write_formula(i, 5, f"=B{i+1}-C{i+1}-D{i+1}-E{i+1}")  # Saldo Final

file_path
```
