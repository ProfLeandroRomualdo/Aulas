```python
# import das bibliotecas

# manipulação de dados e calculo numérico
import pandas as pd
import numpy as np

# visualização de dados
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning - Redução de dimensionalidade
from sklearn.datasets import load_iris, load_wine, fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA, KernelPCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.manifold import TSNE, Isomap, LocallyLinearEmbedding as LLE
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
import umap
from sklearn.datasets import fetch_20newsgroups

# Ignorar avisos
import warnings
warnings.filterwarnings("ignore")
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12
```

```python
iris = load_iris()
X_iris = iris.data
y_iris = iris.target

feature_names = iris.feature_names
target_names = iris.target_names
```

```python
X_iris.shape
```

```python
# 1. Normalização dos dados
scaler = StandardScaler()
X_iris_scaled = scaler.fit_transform(X_iris)
```

```python
# 2. Aplicação do PCA
pca_iris = PCA(n_components=2) # n_components é a quantidade de componentes que vamos reduzir o dataset
X_iris_pca = pca_iris.fit_transform(X_iris_scaled)
```

```python
# 3. Resultados da variância explicada
print(f'Formato dos dados Iris após PCA: {X_iris_pca.shape}')
print(f'Variância explicada por cada componente: {pca_iris.explained_variance_ratio_}')
print(f'Variância total explicada: {np.sum(pca_iris.explained_variance_ratio_):.4f}')
```
