## **Kaplan-Meier**


Kaplan-Meier é um método não paramétrico para estimar a função de sobrevivência em análise de tempo até o evento (sobrevivência), como tempo até a morte, churn de clientes ou falha de equipamento. Ele usa dados censurados (eventos não observados) e calcula a probabilidade cumulativa de sobrevivência S(t) = ∏ [1 - d_i / n_i], onde d_i é o número de eventos no tempo t_i e n_i é o número de indivíduos em risco. A interpretação dos resultados se dá pela curva de sobrevivência: uma linha descendente indica a probabilidade de "sobreviver" ao evento ao longo do tempo, com intervalos de confiança mostrando a incerteza. Valores altos iniciais sugerem baixa taxa de eventos precoces, enquanto quedas acentuadas indicam períodos de alto risco.

## **Log-Rank**

Log-rank é um teste estatístico para comparar curvas de sobrevivência entre grupos (ex.: tratado vs. controle). Ele testa a hipótese nula de que não há diferença entre as funções de sobrevivência, calculando uma estatística qui-quadrada baseada na diferença observada vs. esperada de eventos em cada tempo. A interpretação é pelo p-valor: se <0.05, há diferença significativa; o valor da estatística qui-quadrada indica a força da diferença.

## **Modelos Paramétricos**

Modelos paramétricos assumem uma distribuição específica para o tempo de sobrevivência, como exponencial (risco constante), Weibull (risco variável) ou log-normal. Eles estimam parâmetros da distribuição (ex.: taxa λ na exponencial, onde S(t) = e^{-λt}) usando máxima verossimilhança. A interpretação envolve coeficientes (ex.: hazard ratio) e funções como sobrevivência S(t), risco h(t) e densidade f(t); por exemplo, em Weibull, o parâmetro de forma >1 indica risco crescente ao longo do tempo.

## **COX**

Cox, ou modelo de riscos proporcionais, é semiparamétrico e modela o risco h(t|X) = h0(t) exp(βX), onde h0(t) é o risco base (não especificado) e β são coeficientes das covariáveis X. Assume proporcionalidade de riscos (HR constante). A interpretação foca nos HR = exp(β): HR >1 indica aumento de risco (ex.: HR=1.5 significa 50% mais risco), com intervalos de confiança e p-valores para significância; gráficos de Schoenfeld testam a suposição de proporcionalidade.

## **Coorte**

Coortes referem-se a estudos de coorte, onde um grupo de indivíduos com característica comum é seguido ao longo do tempo para observar a ocorrência de eventos. Em análise de sobrevivência, coortes são usadas para estratificar dados (ex.: coorte de pacientes por idade). A interpretação envolve comparar taxas de eventos entre coortes, usando Kaplan-Meier ou Cox para estimar diferenças; resultados mostram incidência cumulativa ou riscos relativos, ajudando a identificar fatores de risco.

## **Resumo**

Em resumo, esses métodos complementam-se: Kaplan-Meier visualiza sobrevivência, Log-rank compara grupos, paramétricos modelam distribuições, Cox incorpora covariáveis, e coortes fornecem o design de estudo para interpretação contextual. Para resultados, foque em p-valores, HR e curvas para inferências robustas.
