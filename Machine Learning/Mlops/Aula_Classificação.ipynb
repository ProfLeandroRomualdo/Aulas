### **Treinamento do modelo - Abordagem 1**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import SGDClassifier
from xgboost import XGBClassifier

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, recall_score, f1_score, precision_score
```

```python
# Divisão entre treino e teste
X = df.drop('PitNextLap', axis=1)
y = df['PitNextLap']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

#### *Árvore de decisão simples*

```python
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)
print("Acurácia:", accuracy_score(y_test, y_pred))

features = X_train.columns
tree_rules = export_text(clf, feature_names=list(features))
print(tree_rules)
```

```python
from sklearn import tree

cols = list(df.columns)
cols.remove('PitNextLap')

fig = plt.figure(figsize=(15, 15))
p = tree.plot_tree(clf, feature_names=cols, filled=True)
```

#### *Treinamento dos modelos*

```python
lr = LogisticRegression()
dt = DecisionTreeClassifier()
rf = RandomForestClassifier()
knn = KNeighborsClassifier()
nb = GaussianNB()
sgd = SGDClassifier()
xgb = XGBClassifier()
```

```python
models = [lr, dt, rf, knn, nb, sgd, xgb]
d = {}

for i in models:
    i.fit(X_train, y_train)
    y_pred = i.predict(X_test)
    print(i, ':', accuracy_score(y_test, y_pred)*100)
    d.update({str(i):i.score(X_test, y_test)*100})

plt.figure(figsize=(15,5))
plt.title('Acurácia dos modelos')
plt.xlabel('Algoritmo')
plt.ylabel('Acurácia')
plt.plot(d.keys(), d.values(), marker='o', color='plum', linewidth=4, markersize=13,
         markerfacecolor='black', markeredgecolor='black')
plt.show()
```

```python
rf_model = RandomForestClassifier(n_estimators=500,
                                  max_depth=4,
                                  n_jobs=10,
                                  min_samples_split=2,
                                  min_samples_leaf=2,
                                  max_features=None)
rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)
print("Acurácia:", accuracy_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
def plot_shap_explanations(shap_values, X, idx=0, class_idx=1, show_force=True):
    import shap
    import matplotlib.pyplot as plt

    values_shape = shap_values.values.shape
    is_multiclass = len(values_shape) == 3

    if is_multiclass:
        shap_values_global = shap.Explanation(
            values=shap_values.values[:, :, class_idx],
            base_values=shap_values.base_values[:, class_idx],
            data=shap_values.data,
            feature_names=X.columns if hasattr(X, "columns") else None
        )
    else:
        shap_values_global = shap_values

    print("→ Beeswarm")
    shap.plots.beeswarm(shap_values_global, show=True)

    print("→ Bar")
    shap.plots.bar(shap_values_global, show=True)

    if is_multiclass:
        exp = shap.Explanation(
            values=shap_values.values[idx, :, class_idx],
            base_values=shap_values.base_values[idx, class_idx],
            data=shap_values.data[idx],
            feature_names=X.columns if hasattr(X, "columns") else None
        )
    else:
        exp = shap_values[idx]

    print("→ Waterfall")
    shap.plots.waterfall(exp, show=True)

    if show_force:
        print("→ Force plot")

        try:
            shap.initjs()
            force_plot = shap.plots.force(
                exp.base_values,
                exp.values,
                exp.data
            )
            return force_plot  # necessário em notebooks

        except Exception:
            print("Force plot não renderizou (provavelmente fora de notebook).")

            # fallback estático
            shap.force_plot(
                exp.base_values,
                exp.values,
                exp.data,
                matplotlib=True
            )
            plt.show()
```

```python
plot_shap_explanations(shap_values, X_test, idx=0, class_idx=1, show_force=True)
```

#### Leitura e interpretação da explicabilidade

*Importância Global (bar plot)*

Este gráfico apresenta a média absoluta das contribuições de cada variável do modelo.
De acordo com as variáveis no gráfico, o modelo depende fortemente de variáveis relacionadas aos pneus e contexto temporal.


*Distribuição de impacto (beeswarm)*
Neste gráfico temos duas dimensões importantes

- Direção do impacto (positivo | negativo)
- Relação com valor da variável (cor indicativa)

*Principais Insights*

**Tyrelife**
- Valores altos (vermelho) - Aumentam a predição
- Valores baixos (azul) - Reduzem a predição

Relação clara e Monotônica - Não impactam a predição

**Year**
- Anos mais recentes - Impacto positivo
- Anos mais antigos - Impacto negativo

O modelo capturou um efeito temporal que pode se dar por evolução tecnológica, mudança de estratégia ou mudança de regulamento.

**Stint**
- Valores baixos - Impactam negativo forte
- Valores altos - Impactam positivo

Sugere que fases mais avançadas do stint tendem a aumentar a saida do modelo.

**RaceProgress**
- Estranhamente, valores em altos (azul) aparecem com impacto positivo em alguns casos
- Possui distribuição mais espalhada, efeito menos estável

É uma feature com comportamento mais complexo e não linear.





#### **Grafico de dependencia parcial**


```python
from sklearn.inspection import PartialDependenceDisplay
import matplotlib.pyplot as plt

features = ["TyreLife", "Stint", "Year", "RaceProgress"]

fig, ax = plt.subplots(figsize=(10, 6))

PartialDependenceDisplay.from_estimator(
    rf_model,
    X_test,
    features=features,
    kind="average",  # PDP clássico
    ax=ax
)

plt.tight_layout()
plt.show()
```

#### **Shap Interaction Values**

```python
explainer_tree = shap.TreeExplainer(rf_model)

shap_interactions = explainer_tree.shap_interaction_values(X_test)
```

```python
# média absoluta das interações
interaction_matrix = np.abs(shap_interactions).mean(axis=0)

plt.figure(figsize=(15, 15))
plt.imshow(interaction_matrix)
plt.colorbar()

plt.xticks(range(len(X_test.columns)), X_test.columns, rotation=90)
plt.yticks(range(len(X_test.columns)), X_test.columns)

plt.title("SHAP Interaction Strength")
plt.tight_layout()
plt.show()
```
